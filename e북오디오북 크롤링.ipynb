```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import pandas as pd

# 셀레니움 옵션
options = Options()
options.add_argument("--headless")  # ✅ 이 줄이 브라우저 창을 안 띄움
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=Service(), options=options)

# 🔥 "일반 e북" URL로 이동
url = "https://ebook.kyobobook.co.kr/dig/etc/landing/best?lsc=AUD"
driver.get(url)

time.sleep(5)  # 충분히 기다려주자

soup = BeautifulSoup(driver.page_source, 'html.parser')

# 🔥 prodDt 하나하나 가져오기
products = soup.select('div.prodDt')

data = []

for prod in products:
    try:
        # 순위
        rank_tag = prod.select_one('em.rank')
        rank = rank_tag.text.strip() if rank_tag else None
        
        # 제목
        title_tag = prod.select_one('h3 a')
        title = title_tag.text.strip() if title_tag else None
        
        # 상세페이지 링크
        detail_link = title_tag['href'] if title_tag else None
        
        # 저자, 출판사, 출판일 (순서 주의)
        info_spans = prod.select('p.prodDt_info span')
        author = info_spans[0].text.strip() if len(info_spans) > 0 else None
        publisher = info_spans[1].text.strip() if len(info_spans) > 1 else None
        pub_date = info_spans[2].text.strip() if len(info_spans) > 2 else None
        
        # 가격
        price_tag = prod.select_one('div.prodDt_price span b')
        price = price_tag.text.strip() if price_tag else None
        
        data.append({
            '순위': rank,
            '제목': title,
            '상세링크': detail_link,
            '저자': author,
            '출판사': publisher,
            '출판일': pub_date,
            '가격': price
        })
    except Exception as e:
        print("❌ 에러 발생:", e)

driver.quit()

# ✅ 데이터프레임 변환 및 출력
df = pd.DataFrame(data)
print(df)

# CSV 저장
df.to_csv('kyobo_ebook_best.csv', index=False, encoding='utf-8-sig')
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import pandas as pd

# 🧩 셀레니움 기본 설정
options = Options()
options.add_argument("--headless")  # 창 안 뜨게
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=Service(), options=options)

# ✅ 1. 기본 페이지 (일반 e북이 자동 선택됨)
driver.get("https://ebook.kyobobook.co.kr/dig/etc/landing/best")
time.sleep(5)  # JS 로딩 대기

# ✅ 2. BeautifulSoup로 파싱
soup = BeautifulSoup(driver.page_source, 'html.parser')
products = soup.select('div.prodDt')

# ✅ 3. 데이터 파싱
data = []
for prod in products:
    try:
        rank_tag = prod.select_one('em.rank')
        rank = rank_tag.text.strip() if rank_tag else None

        title_tag = prod.select_one('h3 a')
        title = title_tag.text.strip() if title_tag else None
        detail_link = title_tag['href'] if title_tag else None

        info_spans = prod.select('p.prodDt_info span')
        author = info_spans[0].text.strip() if len(info_spans) > 0 else None
        publisher = info_spans[1].text.strip() if len(info_spans) > 1 else None
        pub_date = info_spans[2].text.strip() if len(info_spans) > 2 else None

        price_tag = prod.select_one('div.prodDt_price span b')
        price = price_tag.text.strip() if price_tag else None

        data.append({
            '순위': rank,
            '제목': title,
            '상세링크': detail_link,
            '저자': author,
            '출판사': publisher,
            '출판일': pub_date,
            '가격': price
        })
    except Exception as e:
        print("❌ 에러 발생:", e)

driver.quit()

# ✅ 4. 저장 및 출력
df = pd.DataFrame(data)
print(df)
df.to_csv('kyobo_general_ebook_best.csv', index=False, encoding='utf-8-sig')
```
